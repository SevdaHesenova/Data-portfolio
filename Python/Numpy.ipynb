```python
1
NumPy kitabxanasında əsas obyekt array-dir. Array eyni tipli elementlərdən ibarət massivdir və hesablamalar daha sürətli aparılır.

Bu tapşırıqda:

Python siyahısından (list) NumPy array yaradın

array-in ölçüsünü (shape) ekrana çıxarın

array-in bütün elementlərini 2-yə vurun
```

```python
import numpy as np
list=[1,3,5,7,9]
arr=np.array(list)
print("array:",arr)
print("shape:",np.shape(list))
print("array elements multiplied by 2:",array*2)
```

```python
2
NumPy-də array-lər bir neçə ölçülü ola bilər. İki ölçülü array-lər matrix adlanır və sətir və sütunlardan ibarətdir.

Bu tapşırıqda:

3x3 ölçülü matrix yaradın

matrix-dəki bütün elementlərin cəmini tapın

maksimum elementi müəyyən edin
```

```python
matrix = np.arange(1,10)
matrix.reshape(3,3)
print("total:",matrix.sum())
print("max:",matrix.max())
```

```python
3
NumPy-də müəyyən şərtə uyğun elementləri seçmək mümkündür.

Bu tapşırıqda:

1-dən 10-a qədər array yaradın

yalnız cüt ədədləri seçin və ekrana çıxarın
```

```python
arr = np.arange(1,10)
print("array:",arr)
print("even elements:",arr[arr%2==0])
```

```python
4
NumPy statistik hesablamaları sürətli şəkildə aparmağa imkan verir.
    
Bu tapşırıqda:
    
array yaradın
    
orta qiyməti (mean)
    
standart sapmanı (std)
    
minimum və maksimum qiymətləri hesablayın
```

```python
arr = np.arange(1,10)
print("array:",arr)
print("average:",arr.mean())
print("std:",arr.std())
print("max: ",arr.max())
print("min:",arr.min())
```

```python
5
Bəzən array-in elementləri dəyişmədən yalnız forması dəyişdirilir. Bu əməliyyat reshape adlanır.
    
Bu tapşırıqda:
    
1-dən 12-yə qədər array yaradın
    
array-i 3 sətir və 4 sütundan ibarət matrix formasına salın
```

```python
arr = np.arange(1,13)
print("array:",arr)
a=arr.reshape(3,4)
print("4X3 matrix:",a,sep="\n")
```
