```python
# zip 

n=input("Sozu daxil edin: ")
print("Istifade edeceyimiz soz: ",n)
print("Daxil etdiyimiz sozun olchusu:",len(n))
index=[]
for i in range(0,len(n),2):
    index.append(i)
h=[n[x] for x in index]
zipped=zip(index,h)
for i in zipped:
    print(i)
```
