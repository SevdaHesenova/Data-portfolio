# Task_1  FOR ve WHILE LOOP

# Verilmish list [1,2,3,4,5] daxilindeki ededler 3den boyuk 
# beraberdirse print olunsun.

# 1ci versiya yalniz 3 print olunsun
# 2ci versiya 3 4 5 print olunsun


```python
# 1ci versiya yalniz 3 print olunsun : for
lists=[1,2,3,4,5]
for i in lists:
    if i == 3:
        print(i)
        break
```

```python
# 2ci versiya 3 4 5 print olunsun : for
lists=[1,2,3,4,5]
for i in lists:
    if i == 1 or i == 2:
        continue
    print(i)
        
```

```python
# 1ci versiya yalniz 3 print olunsun : while
lists=[1,2,3,4,5]
a=len(lists)
c=0
while c<a:
    if lists[c]==3:
        print(lists[c])
        break
    c+=1
```

```python
# 2ci versiya 3 4 5 print olunsun : while
lists=[1,2,3,4,5]
a=len(lists)
c=0
while c<a:
    if lists[c]==1 or lists[c]==2:
        c+=1
        continue
    print(lists[c])
    c+=1
```

# Task_2   Funksiya yaradin parametrleri hours, daily income ve bonus olsun

# Gundelik geliri 100 azndir. Eger 8 saatdan chox ishleyerse 25 faiz bomus alacaq
# 8 saatdan az olarsa maashi olduqu kimi qalacaq


```python
def f(h,di,b):
    if h>=8:
        di=di+di*(b/100)
    return di
hours=int(input("is saatini daxil edin: "))
daily_income=int(input("gundelik geliri daxil edin: "))
bonus=int(input("bonusu daxil edin: "))
print(f(h=hours,di=daily_income,b=bonus))
```

# Task_3 - Verilmish ededin daxilindeki reqem sayini gosterecek proqram qurun

# Meselen eded 678-dirse, netice olacaq 3.

```python
# for ile
n=int(input("Eded daxil edin: "))
count=0
for i in str(n):
    count+=1
print(count,"reqemli ededdir")
```
