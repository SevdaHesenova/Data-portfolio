Aşağıda verilmiş stringdə müvafiq tapşırıqları icra edin:

text = "Python, Java; C++ | python | PYTHON"

Tapşırıqlar:
1. Bütün sözləri eyni formata (lowercase) salın
2. Ayırıcıları (",", ";", "|") nəzərə alaraq bütün dilləri ayırın (Ayırıcıları boşluqla əvəz edin)
3. Təkrarlanan sözləri silin
4. Nəticədə cümlə içərisində neçə fərqli (unique) proqramlaşdırma dili olduğunu tapın

İstifadə ediləcək:
lower(), replace(), split(), set(), len() metodları



```python
## Answer
text = "Python,Java;C++|python|PYTHON"
text=text.lower()
print("Lowercase: ",text)
new_text=text.replace(","," ").replace(";"," ").replace("|"," ")
print("Replace: ",new_text)
new_text1=new_text.split(" ")
print("Unique programming languages: ",set(new_text1))
print("Count of unique programming languages: ",len(set(new_text1)))

```

Kiçik bir mağaza müştərinin yaşını daxil edərək bilet qiymətini yoxlamaq üçün sadə proqram istəyir:

Yaş 12-dən kiçikdirsə → Child ticket: 5 dollar

Yaş 12 ilə 17 arasındadırsa (daxil olmaqla) → Teen ticket: 8 dollar

Yaş 18 və ya daha böyükdürsə → Adult ticket: 12 dollar


While loop istifadə edərək istifadəçidən yaşını daxil etməsini xahiş edən proqram yazın. Proqram istifadəçi -1 yazana qədər davam etməlidir.

Daxil edilən hər yaş üçün (yalnız -1 istisna olmaqla) if/elif/else istifadə edərək düzgün bilet qiymətini çap edin.

```python
## Answer
while True:
    age=int(input())
    if age==-1:
        break
    if age<12:
        print("Child ticket: 5 dollar")
        continue
    elif age>=12 and age<=17:
        print("Teen ticket: 8 dollar")
        continue
    elif age>=18:
        print("Adult ticket: 12 dollar")
        continue
        
```

90 və yuxarı → "A"

80–89 → "B"

70–79 → "C"

Funksiyanızı bu siyahı ilə sınayın:

scores = [95, 82, 67, 74, 100, 59]

```python
## Answer
def grade_students(n):
    score=[]
    for i in n:
        if i>=90:
            score.append("A")
        elif i>=80:
            score.append("B")
        elif i>=70:
            score.append("C")
        else:
            score.append("F")
    return score
scores = [95, 82, 67, 74, 100, 59]
'''
n=int(input())
scores=[]
for i in range(n):
    s=int(input())
    scores.append(s)
'''
print(scores)
print(grade_students(scores))
```

sales = [120, 85, 200, 0, 150, 60, 0, 300, 95, 0]

For loop istifadə edərək aşağıdakıları hesablayın və çap edin:

Ümumi satış məbləği (bütün günlərin cəmi)

Satışın olmadığı günlərin sayı (yəni dəyəri 0 olan günlər)

Yalnız satışı 100-dən çox olan günlərin orta (average) qiymətini hesablayın

Qeyd: Hər bir tapşırıq üçün ayrı dəyişən(lər) istifadə edin və nəticələri aydın şəkildə çap edin (məsələn: "Ümumi satış: 1010").

```python
## Answer
sales = [120, 85, 200, 0, 150, 60, 0, 300, 95, 0]
total=0  #Total sales amount
count=0  # Count of non-selling days
s=0 #Total sales of days over 100
c=0  #Number of days with sales over 100
for i in sales:
    total+=i
    if i==0:     #print(sales.count(0))
       count+=1
    if i>=100:
        s+=i
        c+=1
print("Total sales amount: ",total)
print("Count of non-selling days: ",count)
print("Average sales of days over 100: ",s/c)
```
