# TASK 1
 Aşağıdakı DataFrame verilib:

data = {
    "Name": ["Ali", "Veli", "Aysel", "Murad"],
    "Age": [25, 30, 28, 22],
    "Salary": [1200, 1500, 1700, 900]
}

Yaşı 25-dən böyük olanları seçin

Maaşı 1500-dən böyük olanları seçin

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Veli", "Aysel", "Murad"],
    "Age": [25, 30, 28, 22],
    "Salary": [1200, 1500, 1700, 900]
}
df = pd.DataFrame(data)
df["match_age"] = np.where((df.Age >=25), "yes", "no")
df["match_salary"] = np.where((df.Salary >=1500 ), "yes", "no")
df

```

# TASK 3 
Aşağıdakı məlumat verilib:

data = {
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [2000, 2500, 1500, 1800]
}

Hər department üzrə orta maaşı hesablayın.

```python
data = {
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [2000, 2500, 1500, 1800]
}
df = pd.DataFrame(data)
average_salary=df.groupby("Department").mean()
average_salary
```

# TASK 4
DataFrame-də boş dəyərlər var:

data = {
    "Name": ["Ali", "Veli", "Aysel"],
    "Salary": [1200, None, 1700]
}

Boş dəyərləri tapın

Boş maaşı orta maaş ilə doldurun.

```python
data = { 
    "Name": ["Ali", "Veli", "Aysel"], 
    "Salary": [1200, None, 1700] 
}
df = pd.DataFrame(data)
df

df.isnull()
average_salary=df["Salary"].mean()

df["Salary"] = df["Salary"].fillna(average_salary)
df
```
