Temp < 0 və yaş > 65 → "Evə çıxmaq tövsiyə olunmur"

Temp < 0 və xəstəlik=yes və maska=no → "Çox risklidir"

Temp 0–15 → "Ehtiyatlı geyinin"

Temp > 15 və xəstəlik=yes → "Maska taxın və ehtiyatlı olun"

Bonus: Funksiya şəklində yaz:

```python
yash=int(input("Yashınızı daxil edin: "))
temp=int(input("Temperaturunuzu daxil edin: "))
xestelik=input("Xesteliyiniz var? yes/no ")
maska=input("Maska taxırsınız? yes/no ")
if temp < 0:
    if yash >= 65:
        result="Evə çıxmaq tövsiyə olunmur"
    elif xestelik=="yes" and maska=="no":
        result="Çox risklidir"
elif temp>=0 and temp<=15:
    result="Ehtiyatlı geyinin"
else:
    if xestelik=="yes":
        result="Maska taxın və ehtiyatlı olun"
    else:
        result="Problem yoxdur"
print(result)
```

### TASK 3 – İmtahan + Davamiyyət + Davranış + Bonus Ball 
Təsvir:

İstifadəçi:

imtahan balı

davamiyyət faizi

intizam cəzası (yes/no)

əlavə bonus balı

Qaydalar:

Davamiyyət < 75 → "Buraxılmadı"

İmtahan + bonus ≥ 90 → "Əla"

İmtahan ≥ 50 və cəza=yes → "Şərti keçid"

İmtahan ≥ 50 və cəza=no → "Keçdi"

Əks halda → "Kəsildi"

Bonus: Funksiya şəklində yaz:

```python
ib=int(input("imtahan balınızı daxil edin: "))
df=int(input("davamiyyet_faizinizi daxil edin: "))
ic=input("intizam cəzası yes/no olaraq daxil edin: ")
eb=int(input("əlavə bonus balınızı daxil edin: "))
if df<75:
    netice="Buraxılmadı"
else: 
    if ib+eb >= 90:
        netice="Ela"
    elif ib>=50: 
        if ic=="yes":
            netice="Sherti kechid"
        else:
            netice="Kechdi"
    else:
        netice="Kesildi"
print(netice)
```

```python
#Bonus: Funksiya şəklində yaz:
def check(ib,df,ic,eb):
    if df<75:
        netice="Buraxılmadı"
    else: 
        if ib+eb >= 90:
            netice="Ela"
        elif ib>=50 and ic=="yes":
                netice="Sherti kechid"
        elif ib>=50 and ic=="no":
                netice="Kechdi"
        else:
            netice="Kesildi"
    return netice
a=int(input("imtahan balınızı daxil edin: "))
b=int(input("davamiyyet_faizinizi daxil edin: "))
c=input("intizam cəzası yes/no olaraq daxil edin: ")
d=int(input("əlavə bonus balınızı daxil edin: "))
print(check(ib=a,df=b,ic=c,eb=d))
```
