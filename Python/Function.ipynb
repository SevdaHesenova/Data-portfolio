# 4
Limitsiz sayda agrument qebul ede bilen funksiya yaradin, neticede ise sonuncu argument print olunsun.

Misal:

[1,2,3,4,5], ['a', 'b', 'c'], ['x', 'y'] bu arqumentler funksiyaya daxil edilecek lakin netice  ['x', 'y'] olaraq qayidacaq.
    
funksiya_adi([1,2,3,4,5], ['a', 'b', 'c'], ['x', 'y'])

Netice:

['x', 'y']

```python
def lists(*arg):
    d=list(arg)
    return d[len(d)-1]
print(lists([1,2,3,4,5], ['a', 'b', 'c'], ['x', 'y']))
```
